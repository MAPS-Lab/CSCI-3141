# Iris sepal lengths

How do sepal lengths vary within and between the three species in this
sample? Iris contains 150 flowers, 50 of each species, with measurements
in centimetres. These historical example data are not a random sample
of all iris populations.

```python
import platform
from importlib.metadata import version

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import Markdown, display
from sklearn.datasets import load_iris

iris = load_iris()
dat = pd.DataFrame(
    iris.data,
    columns=["Sepal_cm", "SepalWidth_cm", "Petal_cm", "PetalWidth_cm"],
)
dat["Species"] = iris.target_names[iris.target]
species = list(iris.target_names)
```

```python
mean_sepal = dat["Sepal_cm"].mean()
sd_sepal = dat["Sepal_cm"].std(ddof=1)
display(
    Markdown(
        f"The mean sepal length is {mean_sepal:.2f} cm, "
        f"with sample standard deviation {sd_sepal:.3f} cm."
    )
)
```

## A compact summary

```python
summary_table = (
    dat.groupby("Species", sort=True)
    .agg(
        n=("Sepal_cm", "size"),
        Sepal_cm=("Sepal_cm", "mean"),
        Petal_cm=("Petal_cm", "mean"),
    )
    .reset_index()
)
display(
    {
        "text/html": summary_table.to_html(
            index=False, float_format="%.3f", escape=True
        ),
        "text/latex": summary_table.to_latex(
            index=False, float_format="%.3f", escape=True
        ),
    },
    raw=True,
)
```

The table reports means in centimetres. Each species has the same sample
size, but a mean alone does not describe the within-species spread.

## Two views of the variation

```python
edges = np.linspace(4.25, 8.25, 11)
fig, ax = plt.subplots(figsize=(6, 3))
ax.hist(
    dat["Sepal_cm"],
    bins=edges,
    color="#D4A017",
    edgecolor="white",
)
ax.set(xlabel="Sepal length (cm)", ylabel="Number of flowers")
ax.set_title("Figure 1. Sepal length distribution", fontsize=11)
fig.tight_layout()
plt.show()
```

The histogram uses 0.4 cm bins from 4.25 to 8.25 cm and combines species.
The edges avoid the recorded
0.1 cm measurement grid, so no observation lies on an internal edge.

```python
groups = [dat.loc[dat["Species"].eq(name), "Sepal_cm"] for name in species]
fig, ax = plt.subplots(figsize=(6, 3))
ax.boxplot(
    groups,
    tick_labels=species,
    widths=0.55,
    patch_artist=True,
    boxprops={"facecolor": "#D4A017"},
    medianprops={"color": "black"},
)
ax.set(xlabel="Species", ylabel="Sepal length (cm)")
ax.set_title("Figure 2. Sepal lengths by species", fontsize=11)
fig.tight_layout()
plt.show()
```

The boxes use linear quartiles and 1.5-IQR whiskers.
Setosa has the lowest sample median, while the species'
ranges overlap. The boxes reveal variation hidden by the table. They
do not establish that species alone causes a sepal-length difference.
